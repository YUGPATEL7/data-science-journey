# Module 14 – Visualization using Matplotlib 

```python
# Create a bar chart of department-wise students.

import matplotlib.pyplot as plt 

department_counts = df['Department'].value_counts()


department_counts.plot(kind='pie',  hatch=['**O', 'oO', 'O.O', '.||.'],autopct='%1.1f%%',pctdistance=1.10,labeldistance=1.2,figsize=(10,6))

plt.title('Students by Department')
plt.xlabel('Department')
plt.ylabel('Number of Students')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# Create a line chart of average percentage by department
department_counts = df['Department'].value_counts()
dept_avg = df.groupby('Department')['Percentage'].mean()

dept_avg.plot(
    kind='line',
    marker='o',
    color='steelblue',
    figsize=(10, 6)
)

plt.title('Students by Department')
plt.xlabel('Department')
plt.ylabel('Number of Students')
plt.tight_layout()
plt.show()
```

```python
# Create a histogram of percentages.

df['Percentage'].plot(kind='hist', bins=20, color='red',edgecolor = "black" , figsize=(10,6))
plt.title('Students by Percentages')
plt.xlabel('Percentage')
plt.ylabel('Number of Students')
plt.tight_layout()
plt.show()
```

```python
# Create a pie chart for gender distribution.
gender_counts = df['Gender'].value_counts()

gender_counts.plot(
    kind='pie',
    autopct='%1.1f%%',
    colors=['steelblue', 'lightcoral'],
    startangle=90
)
plt.title('Gender Distribution')
plt.tight_layout()
plt.show()
```

```python
# Create a scatter plot between Attendance and Percentage.
plt.figure(figsize=(10, 6))

plt.scatter(
    df['Attendance'],
    df['Percentage'],
    c=df['Attendance'],
    cmap='inferno',
    label='Students'
)
plt.colorbar(label='Attendance')
plt.title('Attendance vs Percentage')
plt.xlabel('Attendance')
plt.ylabel('Percentage')
plt.legend()
plt.show()

```

```python
df['Status'] = df['Percentage'].ge(60).map({
    True: 'Pass',
    False: 'Fail'
})

```

```python
plt.figure(figsize=(10, 6))
counts = df.groupby('Department')['Status'].value_counts().unstack(fill_value=0)

Pass = counts['Pass'].ge(60)
Fail = counts['Fail'].lt(60)

plt.stackplot(
    counts.index,
    Pass.astype(int),
    Fail.astype(int),
    labels=['Pass', 'Fail']
    
)

plt.legend(loc='upper right')
plt.xlabel('Department')
plt.ylabel('Pass / Fail')
plt.grid()
plt.show()
```
