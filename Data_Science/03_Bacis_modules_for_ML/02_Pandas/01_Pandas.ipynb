```python

data["Math"] = data["Math"].astype(int)
data["Maths_Topper"] = data["Math"].apply(
    lambda x: "weak" if x <= 60 else "avg" if x <= 80 else "topper"
)
data
```

# Data Transformation


```python
data.sort_values(["Math"])
```

```python
data.rename(columns = {"Math" : "Maths"} ,inplace=True)
data
```

# MELT 


```python
data.melt(id_vars=["Name"],value_vars=["Maths","Science","English"],var_name="Subject",value_name="Scrore")
```

# Joins 


```python
data.groupby("City")["Maths"].mean()
```

```python
data.groupby(["City","Maths_Topper"])["Percentage"].agg( avg_score="mean", max_salary="max", sum_salary="sum")
```
