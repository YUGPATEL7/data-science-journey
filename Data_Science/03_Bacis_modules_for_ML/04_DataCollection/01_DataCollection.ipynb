```python
import pandas as pd 
from bs4 import BeautifulSoup 
```

```python
allproducts = []
for i in range(1,6):
    f = open(f"D:\data-science-journey\Data_Science\Text_Files\Scrap_html_files\page{i}.html","r")
    html_content=f.read()
    
    soup = BeautifulSoup(html_content,"html.parser")
    product_data  = soup.select("article.product_pod")
    for product in product_data:
        title = product.find("h3").find("a").text
        price  = product.select_one("p.price_color").text.split("£")[1]
        allproducts.append({
            "title":title,
            "price":price
        })

df = pd.DataFrame(allproducts)
df
```
