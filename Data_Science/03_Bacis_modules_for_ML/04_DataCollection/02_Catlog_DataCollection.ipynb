```python

with open(r"D:\data-science-journey\Data_Science\Text_Files\Scrap_html_files\Catlog.html","r",encoding="utf-8") as f:
    html_content = f.read()

f.close()
```

```python
soup = BeautifulSoup(html_content, "html.parser")
```

```python
root = soup.select("li.usa-collection__item")
print(root)
```

```python
dataset_info = []
for data in root:
    title = data.select_one("h3 a").get_text(strip=True)
    organization = data.select_one("ul.usa-collection__meta li a").get_text(strip=True)
    description = data.select_one("p.usa-collection__description").get_text(" ", strip=True)
    format = data.select_one("ul.dataset-resources a").get("data-format")
    Relevance = data.find("div").find("small").text.split()[2]
    views = data.find("div").find("small").text.split()[7]
    

    dataset_info.append({
        "title": title,
        "organization": organization,
        "description": description,
        "fromat":format,
        "Relevance":Relevance,
        "views": views
    })

```

```python
import pandas as pd 

df = pd.DataFrame(dataset_info)
df 
```
