```python
import pandas as pd 
```

```python
df  = pd.read_csv(r"D:\data-science-journey\Data_Science\Data\Flight_Booking_Advanced_Dirty_Dataset_5500_Records.csv")
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce').astype('Int64')
```

```python
# Convert Age to integer
df['TicketPrice'] = df['TicketPrice'].astype(float)
```

```python
# Convert FlightDate to datetime.
df['FlightDate'] = pd.to_datetime(
    df['FlightDate'],
    errors='coerce',
    dayfirst=True
)
```

```python
# Convert CustomerID to string
df['CustomerID'] = df['CustomerID'].astype(str)
```

```python
# Convert PNRNumber to string. 
df['PNRNumber'] = df['PNRNumber'].astype(str)
```

```python
# Convert DepartureTime to time datatype. 
df['DepartureTime'] = pd.to_datetime(df['DepartureTime'], format='%H:%M', errors='coerce').dt.time
```

```python
# Identify columns with object datatype.
object_cols = df.select_dtypes(include='object').columns

print(object_cols)
```

```python
# Convert SeatClass into category. 
df['SeatClass'] = df['SeatClass'].astype('category')
```

```python
# Convert invalid numeric values into NaN. 
df.select_dtypes(include='number').columns.fillna('NaA')
```

```python
df['BookingDate'] = pd.to_datetime(df['BookingDate'], errors='coerce')
```

```python
# Extract month from FlightDate. 
df['FlightDate'].dt.month
```

```python
# Extract year from BookingDate. 
df['BookingDate'].dt.year
```

```python
# Create weekday column
df['Weekday'] = df['FlightDate'].dt.day_name()
```

```python

# Verify datatype conversions. 
print(df.dtypes)
```
