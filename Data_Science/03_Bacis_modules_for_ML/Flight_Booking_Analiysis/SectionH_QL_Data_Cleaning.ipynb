```python
# Write SQL query to count missing EmailID.
df.query("@df['EmailID'].isnull()")
```

```python
# Write SQL query to find duplicate rows. 
df.query("@df.duplicated()")
```

```python
# Write SQL query to remove duplicates
# df.query("@df.drop_duplicates()")
```

```python
# Write SQL query to replace NULL values. 
df['EmailID'] = df['EmailID'].fillna('Null')
```

```python
# Update invalid FlightName
df['FlightName'] = df['FlightName'].replace({
    'Indigo': 'IndiGo',
    'Air india': 'Air India',
    'air india': 'Air India'
})
```

```python
# Remove negative TicketPrice
df = df.query("TicketPrice >= 0")
```

```python
# Count bookings city-wise
city_bookings = df.groupby('City').size()
```

```python
# Count bookings country-wise

df['Country'].value_counts()
```

```python
# Calculate average ticket price
df['TicketPrice'].mean()
```
