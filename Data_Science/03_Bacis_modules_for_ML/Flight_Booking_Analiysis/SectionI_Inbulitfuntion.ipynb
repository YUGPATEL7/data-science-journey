```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

df['Age_Group'] = np.where(
    df['Age'] >= 18,
    'Adult',
    'Minor'
)

df[['Age', 'Age_Group']].head()
```

```python
# np.select()

# Used when you have multiple conditions.
conditions = [
    df['Age'] < 18,
    df['Age'].between(18, 59),
    df['Age'] >= 60
]

choices = [
    'Minor',
    'Adult',
    'Senior'
]

df['Age_Group'] = np.select(
    conditions,
    choices,
    default='Unknown'
)
```
