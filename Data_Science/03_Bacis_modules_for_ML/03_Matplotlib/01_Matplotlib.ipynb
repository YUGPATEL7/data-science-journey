```python
np.random.seed(7) 
plt.figure(figsize=(7, 4.5))
for i in range(50):
    plt.plot(np.random.rand(100), linewidth=1)
plt.title("Too Much Data Can Be Confusing!")
plt.grid(True, alpha=0.4)
plt.tight_layout()
plt.savefig("messy_chart.png", dpi=200)
plt.show()
```

```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]

plt.figure(figsize=(7, 4.5))
plt.plot(years, kohli, color='#F5A623', linestyle='--', linewidth=3, label="Kohli")
plt.plot(years, sehwag, color='#1E824C', linestyle='-.', linewidth=2, label="Sehwag")
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Performance Comparison")
plt.legend()
plt.tight_layout()
# plt.savefig("clean_chart.png", dpi=200)
plt.show()
```

# *Stack Plot* 

```python
days = [1, 2, 3, 4, 5, 6, 7] 
studying = [3, 4, 3, 5, 4, 3, 4]
playing = [2, 2, 1, 1, 2, 3, 2]
watching_tv = [2, 1, 2, 2, 1, 1, 1]
sleeping = [5, 5, 6, 5, 6, 5, 5] 



labels = ['Studying', 'Playing', 'Watching TV', 'Sleeping'] 
colors = ['skyblue', 'lightgreen', 'gold', 'lightcoral'] 

plt.stackplot(days,studying,playing,watching_tv,sleeping,labels=labels,colors=colors)
plt.legend(loc = 'upper right')
plt.xlabel("Days")
plt.ylabel("Hours ")
plt.grid()
plt.show()
```
