# Scatter Plot

```python

study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 45, 50, 55, 60, 65, 75, 85, 90]

# Basic scatter
plt.scatter(study_hours, exam_scores)
plt.title('Study Hours vs Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
```

```python
import numpy as np

scores_normalized = np.array(exam_scores)
sizes = [score * 2 for score in exam_scores]

sc = plt.scatter(study_hours, exam_scores, s=sizes, c=scores_normalized, cmap='viridis')
plt.colorbar(sc, label='Score')
plt.title('Scatter Plot with Colormap')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
```

```python
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 45, 50, 55, 60, 65, 75, 85, 90] 
```

```python
plt.scatter(study_hours,exam_scores)
plt.title("Study Hours VS Scored In Exam")
plt.xlabel("Studyed Hours")
plt.ylabel("Scored In Exam")
plt.show()
```

```python
size = [score *2 for score in exam_scores]
color = ['green' if score >= 60 else "red" for score in exam_scores]
```

```python
plt.scatter(study_hours,exam_scores,s=size, c=color)
plt.title("Study Hours VS Scored In Exam")
plt.xlabel("Studyed Hours")
plt.ylabel("Scored In Exam")
plt.show()
```

## Color Map

```python
import numpy as np 

scores_normlized = np.array(exam_scores)
plt.scatter(study_hours,exam_scores,c=scores_normlized,cmap="inferno")
plt.colorbar(label = "Score")
plt.title("Study Hours VS Scored In Exam")
plt.xlabel("Studyed Hours")
plt.ylabel("Scored In Exam")
plt.show()
```

> Adding Annotations

```python

plt.scatter(study_hours,exam_scores)
plt.scatter(study_hours, exam_scores, c=exam_scores, cmap="inferno")


for i in range(len(exam_scores)):
    plt.annotate(
        f"Student {i + 1}",
        (study_hours[i], exam_scores[i])
    )

plt.colorbar(label="Score")
plt.title("Study Hours VS Scored In Exam")
plt.xlabel("Studyed Hours")
plt.ylabel("Scored In Exam")
plt.show()
```

# seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
tips = sns.load_dataset('tips')  
sns.stripplot(x="day", y="total_bill", data=tips, jitter=True, alpha=0.6)
plt.title("Every individual bill — 244 raw data points")
plt.xlabel("Day")
plt.ylabel("Total Bill ($)")

plt.show()
```

```python
sns.barplot(x="day", y="total_bill", data=tips)
plt.title("sns.barplot() average bill per day, computed automatically")
plt.xlabel("Day")
plt.ylabel("Average Total Bill ($)")

plt.show()
```
