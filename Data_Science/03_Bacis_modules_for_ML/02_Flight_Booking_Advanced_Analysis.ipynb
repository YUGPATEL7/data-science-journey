# Section B: Handling Missing Values

```python
import pandas as pd 
```

```python
df = pd.read_csv(r"D:\data-science-journey\Data_Science\Data\Flight_Booking_Advanced_Dirty_Dataset_5500_Records.csv")
```

```python
# Count null values in CustomerName and EmailID ,MobileNo,FlightDate,SourceCity ,DestinationCity
df[['CustomerName', 'EmailID','MobileNo','FlightDate','SourceCity','DestinationCity']].isnull().sum()
```

```python
# Display rows where EmailID is missing. 
df[df['EmailID'].isnull()]
```

```python
# Replace missing EmailID with "Not Available"
# df['EmailID'] = df['EmailID'].fillna('Not Available')
```

```python
# Replace missing MobileNo with "0000000000".
df['MobileNo'] = df['MobileNo'].fillna(0000000000)
```

```python
# .Replace missing Age with average age. 
# df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
# .Replace missing SourceCity with "Unknown". 
df['SourceCity'] = df['SourceCity'].fillna('Unknown')
```

```python
# .Replace missing DestinationCity with "Unknown". 
df['DestinationCity'] = df['DestinationCity'].fillna('Unknown')
```

```python
# Replace missing FlightDate with current date
from datetime import date

today = date.today()
print(today)  
df['FlightDate'] = df['FlightDate'].fillna(today)
```

```python
# .Remove rows containing missing CustomerName. 
df['CustomerName'] = df['CustomerName'].dropna()
```

```python
# Remove rows containing missing EmailID. 
df['EmailID'] = df['EmailID'].dropna()
df['EmailID'].count()
```

```python
# Display columns with at least 10% missing values
ten_per=df.isnull().mean().round(3)*100
col_name =ten_per[ten_per >= 10]
df[col_name.index]
```

```python
# Fill null Age using forward fill. 

df['Age'] = df['Age'].backfill()
```

```python
# Fill null SeatClass using mode. 

df['SeatClass'] = df['SeatClass'].mode()
```

```python
# .Create indicator column for missing MobileNo. 
df['MobileNo_Missing'] = df['MobileNo'].isnull()
```

```python
# Count rows having multiple missing values.
count = (df.isnull().sum(axis=1) > 1).sum()
count
```

```python
# Display top 20 rows containing maximum missing values. 
(df.isnull().sum(axis=1) > 3).sum()
```

```python
# Create missing value summary table
missing_summary = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Percentage': df.isnull().mean() * 100
})

missing_summary
```

```python
original_df = df.copy()
numeric_cols = df.select_dtypes(include="number").columns

df[numeric_cols] = df[numeric_cols].fillna(
    df[numeric_cols].mean()
)
comparison = pd.DataFrame({
    "Before": original_df.isnull().sum(),
    "After": df.isnull().sum()
})

comparison
```
