# Section D: Data Validation Rules 

```python
# Validate EmailID format. 
import re

pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'

df['Email_Valid'] = df['EmailID'].astype(str).str.match(pattern)

df['Email_Valid']
```

```python
# Validate mobile no. format. 
df['MobileNo_Valid'] = df['MobileNo'].astype(str).str.fullmatch(r'\d{10}')
```

```python
# .Validate CustomerName length. 
df['CustomerName_Valid'] = df['CustomerName'].str.len()>=3
```

```python
# Validate Age range between 18 and 80. 
df['Age_Valid'] = df['Age'].ge(18) & df['Age'].le(80)
```

```python
df['BookingDate_Valid'] = pd.to_datetime(
    df['BookingDate'],
    format='%Y-%m-%d',
    errors='coerce'
).notna()
```

```python
df['FlightNumber_Valid'] = df['FlightNumber'].astype(str).str.fullmatch(r'[A-Za-z]{2}\d{3,4}')
```

```python
df['TicketPrice_Valid'] = df['TicketPrice'].gt(0)
```

```python
valid_status = ['Paid', 'Pending', 'Failed']

df['PaymentStatus_Valid'] = df['PaymentStatus'].isin(valid_status)
```

```python
valid_status = ['Confirmed', 'Cancelled', 'Waiting']

df['BookingStatus_Valid'] = df['BookingStatus'].isin(valid_status)
```
