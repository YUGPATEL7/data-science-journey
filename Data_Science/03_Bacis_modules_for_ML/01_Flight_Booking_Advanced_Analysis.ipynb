```python
# Remove leading and trailing spaces from CustomerName
df['CustomerName'] = df['CustomerName'].str.strip()
```

```python
# FlightName to uperrcase & lowercase 
df['FlightName'].str.upper()
df['FlightName'].str.lower()
```

```python
# Standerdize filght name 
df['FlightName'].str.title()
```

```python
# Identify inconsistent city names. 
df['City'].value_counts()
```

```python
# Correct spelling mistakes in FlightName. 

df["FlightName"].unique()
df['FlightName'].replace({
    "INDIGO ":"IndiGO",
    "indigo":"IndiGO",
    "vistara123":"Vistara",
    " Air India":"Air India"
},inplace=True)
```

```python
# Display records having extra spaces.
df[df.apply(lambda row: row.astype(str).str.endswith(" ").any(), axis=1)]
```

```python
#  Remove duplicate records.
df.drop_duplicates()
```

```python
# Count duplicate records. 
df.drop_duplicates().count()
```

```python
# Find duplicate PNR Numbers
df['PNRNumber'].duplicated().sum()
```

```python
df['PNRNumber'].drop_duplicates()
```

```python
# remove rows containing more than 50% null values. 
df.isnull().mean().round(4)*100
```

```python
# Replace missing Age values using mean
df['Age'] = pd.to_numeric(df['Age'], errors='coerce') # errors='coerce' converts invalid values such as "unknown" or "N/A" into NaN
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
# Replace missing Age values using median
df['Age'] = df['Age'].fillna((df['Age'].median()))
```

```python
# Replace missing SeatClass values using mode. 
df['SeatClass'] = df['SeatClass'].fillna((df['SeatClass'].mode()))
```

```python
# Replace missing PaymentStatus with "Pending".
df['PaymentStatus'].fillna("Pending")
```

```python
# Replace missing FlightDate with BookingDate. 
df['FlightDate'].fillna(df['BookingDate'])
```

```python
# Standardize SeatClass names. 
df['SeatClass'].unique()
```

```python
# Identify invalid PaymentStatus values. 
df['PaymentStatus']
```

```python
df["PaymentStatus"].unique()
```

```python
# Identify invalid Gender values. 
df['Gender'].unique()
```

```python
# Convert all string columns to title case. 
object_col =df.select_dtypes(include="object")
df[object_col.columns] = object_col.apply(lambda col: col.str.title())
```

```python
# Create a cleaned copy of dataset.
cleaned_df = df.copy()
    
```

```python
#  Export cleaned dataset to CSV. 
cleaned_df.to_csv("cleaned_dataset.csv", index=False)      
```
