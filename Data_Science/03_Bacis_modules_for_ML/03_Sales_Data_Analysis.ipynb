```python
# 37.
sales_person = data[:,5]
unique_person = np.unique(data[:,5])
max_count = 0
max_person = ""
total_sales_by_person = {}
for p in unique_person:
    count = np.sum(sales[sales_person == p])
    max_count = count
    max_person = p   
    total_sales_by_person[max_person] = max_count
    print("Sales person name :", max_person)
    print("Number of qty they sales :", max_count)

print("\n\n")
best_sales = max(total_sales_by_person, key=total_sales_by_person.get)
print(f"{best_sales}: {total_sales_by_person[best_sales]}")
# best_sales
```

# � Date-wise / Time Analysis Questions


```python
date = data[:, 1]
sales = data[:, 8].astype(float)

unique_date = np.unique(date)

max_count = 0
max_date = ""

total_sales_by_date = {}

for d in unique_date:

    count = np.sum(sales[date == d])

    total_sales_by_date[d] = count

    if count > max_count:
        max_count = count
        max_date = d

print("Date with Highest Sales :", max_date)
print("Total Sales :", max_count)
```
