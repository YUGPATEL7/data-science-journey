```python
def read_file(filepath):
    student_list = []
    with open(filepath,"r") as f: 
        next(f)
        for line in f:
            row = line.strip().split(",")
            student_list.append(row)
    f.close()
    return student_list
```

# 🏙 City-wise Student Analysis Questions

```python
cities = data[:, 2]

unique_cities = np.unique(cities)

max_count = 0
max_city = ""

for city in unique_cities:
    count = np.sum(cities == city)

    if count > max_count:
        max_count = count
        max_city = city

print("City:", max_city)
print("Number of students:", max_count)
```
